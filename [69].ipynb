### 2. Series 객체 연산

💡 None, Nan, Null 비교
| 종류 | 소속 | 데이터 타입 | 핵심 특징 |
| :--: | :--: | :--: | :--|
| None | 파이썬 (Python) |  | ● 값이 없음 = 비어있음 = 빈칸 |
| Nan | 넘파이(NumPy), 판다스(Pandas) | `float (실수형)` | ● 값이 있음 <br> ● 수학적 계산 불가능한 숫자, 결측치에 에러대신 Nan을 반환 <br> &nbsp;&nbsp;&nbsp;ex) 0 / 0, 무한대 + 무한대, 빈칸
| Null | 데이터베이스 (SQL) |  | ● 값이 없음 = 비어있음 = 빈칸 <br> &nbsp;&nbsp;&nbsp; ⚠️ 파이썬에서 Null을 쓰면 에러가 발생함

### 3. 기술통계량

💡 함수 vs 메서드 구분하기
- 함수 : 단독사용 또는 라이브러리.함수() 
    - (ex) print(), len(), type(), np.array()
- 메서드 : 변수명.메서드()
    - (ex) column.sum()

```python
# ndarray 기능을 그대로 지원
column = Series([188, 92, 45, 63, 79])

print("합계:", column.sum()) # .sum() 메서드
print("평균:", column.mean()) # .mean() 메서드
print("최대:", column.max()) # .max() 메서드
print("최소:", column.min()) # .min() 메서드
print("분산:", column.var()) # .var() 메서드
print("표준편차:", column.std()) # .std() 메서드
print("중앙값:", column.median()) # .median() 메서드
print("1사분위 수:", column.quantile(0.25)) # .quantile() 메서드
print("2사분위 수(중앙값):", column.quantile(0.5)) # .quantile() 메서드
print("3사분위 수:", column.quantile(0.75)) # .quantile() 메서드
```

```python
print(type(column)) 
print(type(column.sum())) # ndarray 기능을 그대로 지원됨을 알 수 있음
print(column.dtype)
```

### 4. 조건 검색

```python
# 샘플 데이터 정의
column = Series(np.arange(0, 100, 10)) # 0 ~ 99까지 10씩 증가하는 값
column
```

```python
# Numpy와 같은 표현식 사용
in1 = column[column > 30]
in1
```

```python
# AND 검색 : 대괄호를 이중으로 표기
in2 = column[column <= 70][column > 10]
in2
```

```python
# OR 검색 : 파이프라인(|)으로 구분
in3 = column[(column <= 10) | (column >= 70)]
in3
```

### 6. 데이터 정렬

```python
# sort_values() 메서드 : 값 기반 정렬 (오름차순)
my_dict = {"c": 456, "a": 123, "b": 789}
origin = Series(my_dict)

sort1 = origin.sort_values()
sort1
```

```python
# sort_values(ascending=False) 메서드 : 값 기반 정렬 (내림차순)
sort2 = origin.sort_values(ascending=False)
sort2
```

```python
# sort_index() 메서드 : 인덱스 기반 정렬 (인덱스 오름차순)
sort3 = origin.sort_index()
sort3
```

```python
# sort_index(ascending=False) 메서드 : 인덱스 기반 정렬 (인덱스 내림차순)
sort4 = origin.sort_index(ascending=False)
sort4
```

## p.18 연습문제 - 점수를 표현하는 Series
🔍 학생의 시험점수를 Pandas의 Series를 사용하여 조건에 맞게 Python코드로 출력하시오
#
- 학생의 시험점수 딕셔너리                                                        
score = {"국어": 85, "영어": 92, "수학": 78, "과학": 95, "사회": 88}
#
- 조건
    - 1. 딕셔너리를 Series객체로 변환
    - 2. 수학 과목의 점수 출력
    - 3. 점수가 90점 이상인 과목만 출력
    - 4. 모든 과목 점수의 평균을 계산하여 출력

```python
import numpy as pd
from pandas import Series

# 1. 딕셔너리를 Series객체로 변환
score = {"국어": 85, "영어": 92, "수학": 78, "과학": 95, "사회": 88}

score_series = Series(score)
print("Series 객체:")
print(score_series)
print('-'*20)

# 2. 수학 과목의 점수 출력
score_math = score_series['수학']
# print("수학 점수:", score_math)
print(f"수학 점수: {score_math}")
print('-'*20)

# 3. 점수가 90점 이상인 과목만 출력
score90 = score_series[score_series >= 90]
print("90점 이상 과목:")
print(score90)
print('-'*20)

# 4. 모든 과목 점수의 평균을 계산하여 출력
score_avg = score_series.mean()
# print('평균:', score_avg)
print(f'평균: {score_avg}')
```
