### 2. 준비작업

💡 Pandas Series 자료형

| 구분 | 기존 자료형 (list, dict, ndarray) | Pandas Series |
| :---: | :---: | :---: |
| 인덱스 존재 방식 | 암묵적으로 존재 | 명시적으로 존재 (list, dict, ndarray를 가공해서 생성) |
| 인덱스의 의미 | 단순히 데이터의 위치를 의미 | 데이터에 부여된 고유한 이름표 |
| 인덱스 데이터 타입 | 오직 숫자 형태 (단순히 위치 표시) | 숫자, 문자열 형태 |
| 성격요기능 | 연속성 자료형 | 인덱스가 명시적으로 존재하는 연속성 자료형 |

```python
# 라이브러리 참조
import numpy as np # numpy라이브러리를 np로 부름
from pandas import Series # pandas라이브러리에서 Series를 가져옴
```

### 3. Series 객체 생성

- Series는 기존의 연속성 자료형을 가공하여 생성된 데이터 구조
- list, ndarray, dict는 인덱스가 암묵적으로 존재하지만 Series는 인덱스가 명시적으로 존재

##### ① 리스트를 통한 Series 객체 생성

```python
# 리스트로 객체 생성 : 인덱스 번호 암묵적으로 존재
my_list = [10,30]
my_list 
```

```python
# 리스트를 통한 Series 객체 생성 : 인덱스 번호 0, 1 명시적으로 존재
s1 = Series(my_list) # Series() 클래스
s1
```

##### ② ndarray를 통한 Series 객체 생성

```python
# ndarray로 객체 생성 : 인덱스 번호 암묵적으로 존재
my_array = np.array([100, 200])
my_array
```

```python
# 리스트를 통한 Series 객체 생성 : 인덱스 번호 0, 1 명시적으로 존재
s2 = Series(my_array)
s2
```

##### ③ 딕셔너리를 통한 Series 객체 생성

```python
# 딕셔너리로 객체 생성 : 인덱스 번호 암묵적으로 존재
my_dict = {"hello": 123, "world": 456, "nice": 789} 
my_dict
```

```python
# 리스트를 통한 Series 객체 생성 : 딕셔너리의 key는 index 역할
s3 = Series(my_dict)
s3
```

### 4. 인덱싱과 슬라이싱

##### ① list나 ndarray로 생성 : 일반적인 연속형 자료형의 경우와 동일

```python
my_series = Series([10, 20, 30, 40, 50])
print("== 인덱싱 ==")
print(my_series[0])

print("== 슬라이싱 ==")
print(my_series[1:3])
```

##### ② 딕셔너리로 생성 : 인덱스가 문자열로 존재

```python
my_series = Series({"a": 100, "b": 200, "c": 300, "d": 400})
print("== 인덱싱 ==")
print(my_series["a"])

print("== 슬라이싱 ==")
print(my_series["b":"d"]) # 인덱스가 문자열인 경우 시작 위치 이상 끝 위치 이하임
```

### 5. Series 값만 추출

```python
# values 프로퍼티 사용 : 값만 추출
my_series = Series({"a": 100, "b": 200, "c": 300, "d": 400})

print(type(my_series.values)) # 데이터 구조(= 객체 타입) : numpy의 ndarray 형식
my_series.values
```

```python
# list() 클래스 사용 : 값을 list 자료형으로 변환
mylist = list(my_series.values)

print(type(mylist)) # 데이터 구조(= 객체 타입) : list 형식
mylist
```

### 6. Series 인덱스만 추출

```python
# index 프로퍼티 사용
my_series = Series({"a": 100, "b": 200, "c": 300, "d": 400})

print(type(my_series.index)) # 데이터 구조(= 객체 타입) : 판다스의 인덱스 형식
my_series.index
```

```python
# index를 list 타입으로 변환
mylist = list(my_series.index)

print(type(mylist)) # 데이터 구조(= 객체 타입) : list 형식
mylist
```
