### 3. DataFrame 생성

##### ① 2차원 리스트 → DataFrame 변환

- 인덱스는 0부터 일련번호로 자동 부여됨

```python
# 어느 학급의 성적을 표현한 2차원 리스트
grade_data = [
                [1, "남자", 98,  77,  88,  64],
                [2, "여자", 88,  90,  62,  72],
                [1, "남자", 92,  70,  83,  79],
                [3, "여자", 63,  60,  31,  70],
                [4, "남자", 75,  50,  90,  88]
            ]

# 2차원 리스트를 데이터프레임으로 변환
df = DataFrame(grade_data)
df
```

```python
# 인덱스와 컬럼 이름 지정
i_names = ['철수', '영희', '민철', '수현', '호영'] # 인덱스 이름으로 사용할 리스트
c_names = ['학년', '성별', '국어', '영어', '수학', '과학'] # 컬럼 이름으로 사용할 리스트

# 인덱스와 컬럼이름 지정하기
df = DataFrame(grade_data, index=i_names, columns=c_names)
df
```

##### ② 리스트를 원소로 갖는 딕셔너리 → DataFrame 변환

- 인덱스는 0부터 일련번호로 자동 부여됨
- Series에서는 Key가 index가 됨
- DataFrame에서는 딕셔너리 key가 컬럼(열)이름이 됨

```python
# 어느 학급의 성적표를 표현한 딕셔너리
grade_dic = {
    '학년': [1, 2, 1, 3, 4],
    '성별': ['남자', '여자', '남자', '여자', '남자'],
    '국어': [98, 88, 92, 63, 75],
    '영어': [77, 90, 70, 60, 50],
    '수학': [88, 62, 83, 31, 90],
    '과학': [64, 72, 79, 70, 88]
}

df = DataFrame(grade_dic) # key는 컬럼(열)이 됨
df
```

```python
# 인덱스와 인덱스 이름을 지정한 DataFrame
df = DataFrame(grade_dic,
               index=['철수', '영희', '민철', '수현', '호영']) # 인덱스 지정

# 인덱스 이름 지정하기
df.index.name = '이름'
df
```

### 4. 인덱스 다루기

##### ① set_index('컬럼') 메서드 : DataFrame의 특정 컬럼을 index로 지정 (기존 인덱스 → 지정한 컬럼 인덱스로 변경)

```python
# 원본 데이터 프레임 생성 (리스트를 원소로 갖는 딕셔너리 → DataFrame 변환)
grade = {
    "이름": ['철수', '영희', '민철', '수현', '호영'],
    "학년": [1, 2, 1, 3, 4],
    "성별": ['남자', '여자', '남자', '여자', '남자'],
    "국어": [98, 88, 92, 63, 75],
    "영어": [77, 90, 70, 60, 50],
    "수학": [88, 62, 83, 31, 90],
    "과학": [64, 72, 79, 70, 88]
    }

df = DataFrame(grade)
df
```

- set_index('컬럼') 메서드 : 원본은 변화 없고, 변경사항이 저장된 `복사본 리턴`

```python
# set_index('컬럼') 메서드 : DataFrame의 특정 컬럼을 index로 지정하고 복사본 리턴
df_setindex = df.set_index('이름')
df_setindex
```

```python
# df_setindex 복사본을 리턴하여 원본 df는 변화 없음
df
```

- set_index('컬럼', inplace = True) 메서드 : 복사본 리턴 없이 원본에 바로 적용 <br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; → `원본데이터를 변경하는건 위험하기 때문에 권장 X`

```python
# set_index('컬럼', inplace=True) 메서드 : 복사본 리턴 없이 원본에 바로 적용
df_setindex2 = df.set_index('이름', inplace=True)
df_setindex2
```

```python
# df_setindex2 복사본 리턴없이 원본에 적용하여 df는 변화 있음
df 
```

##### ② reset_index() 메서드 : 인덱스 해제 (인덱스를 일반 컬럼으로 되돌림)

```python
# 원본 데이터 프레임
df
```

- reset_index() 메서드 
    - 인덱스를 컬럼으로 되돌린 후 일련번호 형식의 인덱스 새롭게 생성한 `복사본 리턴`하여 원본 변화 없음

```python
# reset_index() : 인덱스 해제 후 복사본 리턴
df_reset = df.reset_index()
df_reset
```

```python
# df_reset 복사본을 리턴하여 원본 df는 변화 없음
df 
```

- reset_index(inplacr=True) 메서드 : 복사본 리턴 없이 원본에 바로 적용 <br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; → `원본데이터를 변경하는건 위험하기 때문에 권장 X`

```python
# index_reset(inplace=True) 메서드 : 복사본 리턴 없이 원본에 바로 적용
df_reset2 = df.reset_index(inplace=True)
df_reset2
```

```python
# df_reset2 복사본 리턴없이 원본에 적용하여 df는 변화 있음
df 
```

- reset_index() 사용시 주의사항
    - 인덱스를 컬럼으로 되돌린 후 일련번호 형식의 인덱스 새롭게 생성하기 때문에 일려번호 인덱스가 중복

```python
df_reset3 = df.reset_index()
df_reset3
```

- reset_index(drop=True) : 기존 인덱스를 제거하고, 일련번호 형식의 인덱스 새롭게 생성

```python
# 원본데이터
df
```

```python
# '이름'컬럼을 인덱스로 변경하기
df2 = df.set_index('이름')
df2
```

```python
# reset_index(drop=True) : 기존 인덱스 제거하고 일련번호 형식 인덱스 새롭게 생성
df3 = df2.reset_index(drop=True)
df3 # '이름'컬럼이 삭제됨
```

### 5. 외부 파일로부터 데이터 프레임 생성

① CSV 파일을 데이터 프레임으로 가져오기

- pandas의 `read_csv()`함수를 사용
- 로컬에 위치한 파일을 상대,절대 경로 방식으로 가져올 수 있으며 온라인상의 파일을 URL을 기반으로 가져올 수 있음
- CSV파일의 첫 행이 데이터 프레임의 컬럼 이름으로 사용됨
- 인코딩 필요함

```python
# pandas의 read_csv() 함수를 사용하여 데이터 로드
df = read_csv("https://data.juyounga.kr/data/grade.csv",
              encoding="euc-kr") # 인코딩
df
```

```python
# index_col = '컬럼' : 인덱스로 사용할 열 이름 지정
df = read_csv("https://data.juyounga.kr/data/grade.csv",
              encoding="euc-kr", # 인코딩
              index_col='이름')
df
```

② xlsx 파일을 데이터 프레임으로 가져오기
- 인코딩 필요X

```python
# index_col = '컬럼' : 인덱스로 사용할 열 이름 지정
df = read_excel("https://data.juyounga.kr/data/grade.xlsx",
                index_col='이름') # 인코딩 필요 X
df
```
